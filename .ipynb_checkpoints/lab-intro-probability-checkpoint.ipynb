**Challenge 1**

#### Ironhack Airlines 

Often Airlines sell more tickets than they have seats available, this is called overbooking. Consider the following:
- A plane has 450 seats. 
- Based on historical data we conclude that each individual passenger has a 3% chance of missing it's flight. 

If the Ironhack Airlines routinely sells 460 tickets, what is the chance that they have a seats for all passenger?

```python
from scipy.stats import binom

total_tickets = 460
total_seats = 450
prob_missing = 0.03
prob_showing_up = 1 - prob_missing


probability = binom.cdf(total_seats, total_tickets, prob_showing_up)

print(f"The probability that Ironhack Airlines has enough seats is: {probability:.4f}")
print(f"In percentage terms: {probability * 100:.2f}%")
```

**Challenge 2**

#### Ironhack Call Center 

Suppose a customer service representative at a call center is handling customer complaints. Consider the following:
- The probability of successfully resolving a customer complaint on the first attempt is 0.3. 


What is the probability that the representative needs to make at least three attempts before successfully resolving a customer complaint?

```python
from scipy.stats import geom

p = 0.3

probability = geom.sf(2, p)

print(f"The probability of needing at least 3 attempts is: {probability:.4f}")
print(f"In percentage terms: {probability * 100:.2f}%")
```

**Challenge 3**

#### Ironhack Website

Consider a scenario related to Ironhack website traffic. Where:
- our website takes on average 500 visits per hour.
- the website's server is designed to handle up to 550 vists per hour.


What is the probability of the website server being overwhelmed?

```python
from scipy.stats import poisson

avg_visits = 500

max_capacity = 550

probability = poisson.sf(max_capacity, avg_visits)

print(f"The probability of the server being overwhelmed is: {probability:.4f}")
print(f"In percentage terms: {probability * 100:.2f}%")
```

What is the probability of being overwhelmed at some point during a day? (consider 24hours)

```python
mu = 500
limit = 550

# 1. Probability of being overwhelmed in a single hour
p_hour = poisson.sf(limit, mu)

# 2. Probability of being overwhelmed at least once in 24 hours
# Using the complement: 1 - P(Not overwhelmed in any of the 24 hours)
p_day = 1 - (1 - p_hour)**24

print(f"Probability of being overwhelmed in a single hour: {p_hour:.4f}")
print(f"Probability of being overwhelmed at least once in 24 hours: {p_day:.4f}")
print(f"In percentage terms: {p_day * 100:.2f}%")
```

**Challenge 4**

#### Ironhack Helpdesk

Consider a scenario related to the time between arrivals of customers at a service desk.

On average, a customers arrives every 10minutes.

What is the probability that the next customer will arrive within the next 5 minutes?

```python
from scipy.stats import expon
mean_arrival = 10 

# We want the probability of arrival within 5 minutes
target_time = 5

# Calculate P(X <= 5)
# expon.cdf(x, scale=mean)
probability = expon.cdf(target_time, scale=mean_arrival)

print(f"The probability of a customer arriving within 5 minutes is: {probability:.4f}")
print(f"In percentage terms: {probability * 100:.2f}%")
```

If there is no customer for 15minutes, employees can that a 5minutes break.

What is the probability an employee taking a break?

```python
from scipy.stats import expon

mean_arrival = 10

probability_break = expon.sf(15, scale=mean_arrival)

print(f"The probability of an employee taking a break is: {probability_break:.4f}")
print(f"In percentage terms: {probability_break * 100:.2f}%")
```

**Challenge 5**

The weights of a certain species of birds follow a normal distribution with a mean weight of 150 grams and a standard deviation of 10 grams. 

- If we randomly select a bird, what is the probability that its weight is between 140 and 160 grams?

```python
from scipy.stats import norm

mean_weight = 150
std_deviation = 10

prob_between = norm.cdf(160, loc=mean_weight, scale=std_deviation) - \
               norm.cdf(140, loc=mean_weight, scale=std_deviation)

print(f"The probability that the weight is between 140 and 160g is: {prob_between:.4f}")
print(f"In percentage terms: {prob_between * 100:.2f}%")
```

**Challenge 6**

If the lifetime (in hours) of a certain electronic component follows an exponential distribution with a mean lifetime of 50 hours, what is the probability that the component fails within the first 30 hours?

```python
from scipy.stats import expon

mean_lifetime = 50

time_limit = 30

probability = expon.cdf(time_limit, scale=mean_lifetime)

print(f"The probability that the component fails within 30 hours is: {probability:.4f}")
print(f"In percentage terms: {probability * 100:.2f}%")
```
