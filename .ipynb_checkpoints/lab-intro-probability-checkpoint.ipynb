**Challenge 1**

#### Ironhack Airlines 

Often Airlines sell more tickets than they have seats available, this is called overbooking. Consider the following:
- A plane has 450 seats. 
- Based on historical data we conclude that each individual passenger has a 3% chance of missing it's flight. 

If the Ironhack Airlines routinely sells 460 tickets, what is the chance that they have a seats for all passenger?

```python
# Step 1: Import the required module
import scipy.stats as stats

# Step 2: Define the variables
n = 460  # number of passengers 
p = 0.97  # probability that a passenger shows up
k = 450  # maximum number of seats

# Step 3: Calculate the cumulative probability P(X <= 450)
probability = stats.binom.cdf(k, n, p)

print(probability)
```

**Challenge 2**

#### Ironhack Call Center 

Suppose a customer service representative at a call center is handling customer complaints. Consider the following:
- The probability of successfully resolving a customer complaint on the first attempt is 0.3. 


What is the probability that the representative needs to make at least three attempts before successfully resolving a customer complaint?

```python
# Step 1: Define the probability of success
p = 0.3  # Probability of successfully resolving the complaint on an attempt

# Step 2: Calculate the probabilities for 1st and 2nd attempts
P_X_1 = p  # Probability of resolving on the 1st attempt
P_X_2 = (1 - p) * p  # Probability of failing the 1st attempt and resolving on the 2nd

# Step 3: Calculate the probability of at least 3 attempts
P_X_ge_3 = 1 - (P_X_1 + P_X_2)

print(P_X_ge_3)
```

**Challenge 3**

#### Ironhack Website

Consider a scenario related to Ironhack website traffic. Where:
- our website takes on average 500 visits per hour.
- the website's server is designed to handle up to 550 vists per hour.


What is the probability of the website server being overwhelmed?

```python
# Step 1: Import the required module
import scipy.stats as stats

# Step 2: Define the average rate (lambda) and the threshold for the server limit
lambda_rate = 500  # average visits per hour
threshold = 550  # server limit

# Step 3: Calculate the cumulative probability P(X <= 550)
probability_up_to_550 = stats.poisson.cdf(threshold, lambda_rate)

# Step 4: Calculate the probability of more than 550 visits
probability_overwhelmed = 1 - probability_up_to_550

print(probability_overwhelmed)
```

**Challenge 4**

#### Ironhack Helpdesk

Consider a scenario related to the time between arrivals of customers at a service desk.

On average, a customers arrives every 10minutes.

What is the probability that the next customer will arrive within the next 5 minutes?

```python
import math

#Define the average time between arrivals and the time we're interested in
lambda_rate = 1 / 10  # customer arrives every 10 minutes, so lambda = 1/10
time = 5  # time in minutes for which we want the probability

# Calculate the CDF using the exponential distribution formula
probability = 1 - math.exp(-lambda_rate * time)

print(probability)
```

If there is no customer for 15minutes, employees can that a 5minutes break.

What is the probability an employee taking a break?

```python
import math

#Define the rate parameter (lambda) and the time
lambda_rate = 1 / 10  # Average arrival rate is 1 customer per 10 minutes
time = 15  # Time in minutes (we want no customer for 15 minutes)

#Calculate the probability that no customer arrives in 15 minutes
probability = math.exp(-lambda_rate * time)

#Print the result
print(probability)
```

**Challenge 5**

The weights of a certain species of birds follow a normal distribution with a mean weight of 150 grams and a standard deviation of 10 grams. 

- If we randomly select a bird, what is the probability that its weight is between 140 and 160 grams?

```python
import scipy.stats as stats

#Define the mean and standard deviation
mean = 150  # Mean weight in grams
std_dev = 10  # Standard deviation in grams

#Calculate the CDF for 160 grams and 140 grams
prob_160 = stats.norm.cdf(160, mean, std_dev)  # P(X <= 160)
prob_140 = stats.norm.cdf(140, mean, std_dev)  # P(X <= 140)

#Calculate the probability that the weight is between 140 and 160 grams
prob_between_140_and_160 = prob_160 - prob_140

#Print the result
print(prob_between_140_and_160)
```

**Challenge 6**

If the lifetime (in hours) of a certain electronic component follows an exponential distribution with a mean lifetime of 50 hours, what is the probability that the component fails within the first 30 hours?

```python
import math

#Define the rate parameter (lambda) and the time
mean_lifetime = 50  # mean lifetime in hours
lambda_rate = 1 / mean_lifetime  # rate parameter
time = 30  # time in hours for which we want the probability

#Calculate the CDF using the exponential distribution formula
probability = 1 - math.exp(-lambda_rate * time)

print(probability)
```
