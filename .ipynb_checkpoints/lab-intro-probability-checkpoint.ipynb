**Challenge 1**

#### Ironhack Airlines 

Often Airlines sell more tickets than they have seats available, this is called overbooking. Consider the following:
- A plane has 450 seats. 
- Based on historical data we conclude that each individual passenger has a 3% chance of missing it's flight. 

If the Ironhack Airlines routinely sells 460 tickets, what is the chance that they have a seats for all passenger?

```python
from scipy.stats import binom

n_tickets = 460
p_show_up = 0.97
plane_seats = 450

# We need the probability that 450 or fewer passengers show up.
prob_seats_for_all = binom.cdf(plane_seats, n=n_tickets, p=p_show_up)

print(f"Probability that there are seats for all passengers: {prob_seats_for_all:.4f}")
print(f"As a percentage: {prob_seats_for_all * 100:.2f}%")
```

**Challenge 2**

#### Ironhack Call Center 

Suppose a customer service representative at a call center is handling customer complaints. Consider the following:
- The probability of successfully resolving a customer complaint on the first attempt is 0.3. 


What is the probability that the representative needs to make at least three attempts before successfully resolving a customer complaint?

```python
from scipy.stats import geom

p_success = 0.3

prob_at_least_3_attempts = 1 - geom.cdf(2, p=p_success)

print(f"Probability of needing at least 3 attempts: {prob_at_least_3_attempts:.4f}")
print(f"As a percentage: {prob_at_least_3_attempts * 100:.2f}%")
```

**Challenge 3**

#### Ironhack Website

Consider a scenario related to Ironhack website traffic. Where:
- our website takes on average 500 visits per hour.
- the website's server is designed to handle up to 550 vists per hour.


What is the probability of the website server being overwhelmed?

```python
from scipy.stats import poisson

avg_visits_per_hour = 500
server_capacity = 550

prob_overwhelmed_hour = 1 - poisson.cdf(server_capacity, mu=avg_visits_per_hour)

print(f"Probability of server being overwhelmed in one hour: {prob_overwhelmed_hour:.4f}")
print(f"As a percentage: {prob_overwhelmed_hour * 100:.2f}%")
```

What is the probability of being overwhelmed at some point during a day? (consider 24hours)

```python
# Probability of being overwhelmed at least once during a day
# Assuming each hour is independent:
# P(at least one overwhelmed hour) = 1 - P(no overwhelmed hours in 24 hours)

hours_per_day = 24

prob_overwhelmed_day = 1 - (1 - prob_overwhelmed_hour) ** hours_per_day

print(f"Probability of being overwhelmed at some point during a day: {prob_overwhelmed_day:.4f}")
print(f"As a percentage: {prob_overwhelmed_day * 100:.2f}%")
```

**Challenge 4**

#### Ironhack Helpdesk

Consider a scenario related to the time between arrivals of customers at a service desk.

On average, a customers arrives every 10minutes.

What is the probability that the next customer will arrive within the next 5 minutes?

```python
# Challenge 4: Ironhack Helpdesk
# Time between arrivals can be modeled using an exponential distribution.
# Mean waiting time = 10 minutes.
# We want P(next customer arrives within 5 minutes).

from scipy.stats import expon

mean_waiting_time = 10

prob_arrives_within_5 = expon.cdf(5, scale=mean_waiting_time)

print(f"Probability that the next customer arrives within 5 minutes: {prob_arrives_within_5:.4f}")
print(f"As a percentage: {prob_arrives_within_5 * 100:.2f}%")
```

If there is no customer for 15minutes, employees can that a 5minutes break.

What is the probability an employee taking a break?

```python
# Employees can take a break if no customer arrives for 15 minutes.
# That means we need P(T > 15).
# We use survival function: sf(x) = 1 - cdf(x)

prob_break = expon.sf(15, scale=mean_waiting_time)

print(f"Probability that an employee can take a break: {prob_break:.4f}")
print(f"As a percentage: {prob_break * 100:.2f}%")
```

**Challenge 5**

The weights of a certain species of birds follow a normal distribution with a mean weight of 150 grams and a standard deviation of 10 grams. 

- If we randomly select a bird, what is the probability that its weight is between 140 and 160 grams?

```python
# Mean = 150g, standard deviation = 10g.
# We want P(140 < X < 160).

from scipy.stats import norm

mean_weight = 150
std_weight = 10

prob_between_140_160 = norm.cdf(160, loc=mean_weight, scale=std_weight) - norm.cdf(140, loc=mean_weight, scale=std_weight)

print(f"Probability that bird weight is between 140g and 160g: {prob_between_140_160:.4f}")
print(f"As a percentage: {prob_between_140_160 * 100:.2f}%")
```

**Challenge 6**

If the lifetime (in hours) of a certain electronic component follows an exponential distribution with a mean lifetime of 50 hours, what is the probability that the component fails within the first 30 hours?

```python
# Mean lifetime = 50 hours.
# probability that it fails within the first 30 hours: P(T <= 30).

from scipy.stats import expon

mean_lifetime = 50

prob_fails_within_30 = expon.cdf(30, scale=mean_lifetime)

print(f"Probability that the component fails within the first 30 hours: {prob_fails_within_30:.4f}")
print(f"As a percentage: {prob_fails_within_30 * 100:.2f}%")
```
