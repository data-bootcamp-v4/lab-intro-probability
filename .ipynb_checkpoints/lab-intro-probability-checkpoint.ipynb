**Challenge 1**

#### Ironhack Airlines 

Often Airlines sell more tickets than they have seats available, this is called overbooking. Consider the following:
- A plane has 450 seats. 
- Based on historical data we conclude that each individual passenger has a 3% chance of missing it's flight. 


If the Ironhack Airlines routinely sells 460 tickets, what is the chance that they have a seats for all passenger?

```python
from scipy.stats import binom

# k = 460 - 450 = 10

n = 460
p = 0.03
k = 10

binom_dist = binom(n,p)
probability_enough_seats = 1 - binom_dist.cdf(k - 1)

print(f"The probability of having seats for all passengers is {probability_enough_seats}")
```

**Challenge 2**

#### Ironhack Call Center 

Suppose a customer service representative at a call center is handling customer complaints. Consider the following:
- The probability of successfully resolving a customer complaint on the first attempt is 0.3. 


What is the probability that the representative needs to make at least three attempts before successfully resolving a customer complaint?

```python
from scipy.stats import geom

p = 0.3
geom_dist = geom(p)

probability_at_least_three = 1 - geom_dist.cdf(2)

print(f"The probability of needing at least three attempts is {probability_at_least_three}")
```

**Challenge 3**

#### Ironhack Website

Consider a scenario related to Ironhack website traffic. Where:
- our website takes on average 500 visits per hour.
- the website's server is designed to handle up to 550 vists per hour.


What is the probability of the website server being overwhelmed?

```python
from scipy.stats import poisson

mu = 500
limit = 550
poisson_dist = poisson(mu)
probability_overwhelmed = 1 - poisson_dist.cdf(limit)

print(f"The probability that the website is overwhelmed is {probability_overwhelmed}")
```

What is the probability of being overwhelmed at some point during a day? (consider 24hours)

```python
import numpy as np

mu = 500
limit = 550
hours_in_day = 24

poisson_dist = poisson(mu)

p_overwhelmed_hourly = 1 - poisson_dist.cdf(limit)

p_not_overwhelmed_hourly = 1 - p_overwhelmed_hourly

p_not_overwhelmed_daily = p_not_overwhelmed_hourly ** hours_in_day

probability_overwhelmed_daily = 1 - p_not_overwhelmed_daily

print(f"The probability of the server being overwhelmed at some point during the day is: {probability_overwhelmed_daily}")
```

```python
import numpy as np

mu = 500
limit = 550
hours_in_day = 24

poisson_dist = poisson(mu)

p_overwhelmed_hourly = 1 - poisson_dist.cdf(limit) ** hours_in_day

print(f"The probability of the server being overwhelmed at some point during the day is: {p_overwhelmed_hourly}")
```

**Challenge 4**

#### Ironhack Helpdesk

Consider a scenario related to the time between arrivals of customers at a service desk.

On average, a customers arrives every 10minutes.

What is the probability that the next customer will arrive within the next 5 minutes?

```python
from scipy.stats import expon

mean_time = 10
rate = 1 / mean_time 

time_limit = 5

lambda_inv = expon(scale = 1/rate)

probability_within_5_min = lambda_inv.cdf(time_limit)

print(f"The probability that the next customer arrives within 5 minutes is: {probability_within_5_min}")
```

If there is no customer for 15minutes, employees can that a 5minutes break.

What is the probability an employee taking a break?

```python
break_time = 15

lambda_inv = expon(scale = 1/rate)

probability_of_break = 1 - lambda_inv.cdf(break_time)

print(f"The probability that an employee takes a break is: {probability_of_break}")
```

**Challenge 5**

The weights of a certain species of birds follow a normal distribution with a mean weight of 150 grams and a standard deviation of 10 grams. 

- If we randomly select a bird, what is the probability that its weight is between 140 and 160 grams?

```python
from scipy.stats import norm 

mean = 150
std = 10

lower_bound = 140
upper_bound = 160

norm_dist = norm(loc = mean, scale = std)

probability_between_bounds = norm_dist.cdf(upper_bound) - norm_dist.cdf(lower_bound)

print(f"The probability that the bird's weight is between 140 and 160 grams is: {probability_between_bounds}")
```

**Challenge 6**

If the lifetime (in hours) of a certain electronic component follows an exponential distribution with a mean lifetime of 50 hours, what is the probability that the component fails within the first 30 hours?

```python
from scipy.stats import expon

mean_time = 50
rate = 1 / mean_time

time_limit = 30

lambda_inv = expon(scale = mean_time)

probability_failure_within_30_hrs = lambda_inv.cdf(time_limit)

print(f"The probability that the component fails within the first 30 hours is: {probability_failure_within_30_hrs}")
```
