## challenge 3

```python
from scipy.stats import poisson

lam = 500 #average visits per hour 
capacity = 550

prob_overwhelmed = 1 - poisson.cdf(capacity, lam)

print(f"Probability server is orverwhelmed: {prob_overwhelmed:.4f}")
```

```python
# Parameters
lam = 500       # average visits per hour
capacity = 550
hours = 24

# Probability overwhelmed in 1 hour
p_hour = 1 - poisson.cdf(capacity, lam)

# Probability overwhelmed at least once in a day
p_day = 1 - (1 - p_hour)**hours

print(f"Probability server overwhelmed at least once in a day: {p_day:.4f}")
```

## challenge 4

```python
from scipy.stats import expon
import math
mean_interarrival = 10
lam = 1 / mean_interarrival
t = 5

#unsing scipy
prob = expon.cdf(t, scale=1/lam)

#manual formula
prob_manual = 1 - math.exp(-lam * t)

print(f"Probability next customer arrives within 5 minutes: {prob:.4f}")
print(f"(manual check) {prob_manual:.4f}")
```

```python
# Parameters
mean_interarrival = 10   # minutes
lam = 1 / mean_interarrival
t = 15  # minutes with no customers

# Probability no arrival in 15 min
prob_break = math.exp(-lam * t)

print(f"Probability employee can take a break: {prob_break:.4f}")
```

```python
from scipy.stats import norm
mu = 150 
sigma = 10

prob = norm.cdf(160, mu, sigma) - norm.cdf(140, mu, sigma)

print(f"Probability birrd weighs between 140g and 160g: {prob:.4f}")
```
