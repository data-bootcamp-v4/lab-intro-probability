**Challenge 1**

#### Ironhack Airlines 

Often Airlines sell more tickets than they have seats available, this is called overbooking. Consider the following:
- A plane has 450 seats. 
- Based on historical data we conclude that each individual passenger has a 3% chance of missing it's flight. 

If the Ironhack Airlines routinely sells 460 tickets, what is the chance that they have a seats for all passenger?

```python
#code here
```

```python
import scipy.stats as stats

#Given values
n = 460  # Number of tickets sold
p = 0.97  # Probability a passenger shows up
mean = n * p  # Mean of the binomial distribution
std_dev = (n * p * (1 - p))**0.5  # Standard deviation of the binomial distribution

# X value (number of passengers that can be seated)
X = 450

# Calculate Z-score
z_score = (X - mean) / std_dev

# Calculate the probability using the cumulative distribution function (CDF)
probability = stats.norm.cdf(z_score)

print(probability)  # Probability that 450 or fewer passengers show up
```

#### Ironhack Call Center 

Suppose a customer service representative at a call center is handling customer complaints. Consider the following:
- The probability of successfully resolving a customer complaint on the first attempt is 0.3. 


What is the probability that the representative needs to make at least three attempts before successfully resolving a customer complaint?

```python
#code here
```

```python
# Given probability of success and failure
p_success = 0.3  # Probability of success on the first attempt
p_failure = 1 - p_success  # Probability of failure on the first attempt

# Probability that the representative needs at least three attempts
probability_at_least_three_attempts = p_failure * p_failure

print(probability_at_least_three_attempts)  # Probability that the representative needs at least three attempts
```

**Challenge 3**

#### Ironhack Website

Consider a scenario related to Ironhack website traffic. Where:
- our website takes on average 500 visits per hour.
- the website's server is designed to handle up to 550 vists per hour.


What is the probability of the website server being overwhelmed?

```python
#code here
```

```python
import scipy.stats as stats

# Given values
lambda_visits = 500  # Average visits per hour
max_visits = 550  # Threshold where server gets overwhelmed

# Calculate the probability of the server being overwhelmed (i.e., more than 550 visits)
probability_overwhelmed = 1 - stats.poisson.cdf(max_visits, lambda_visits)

print(probability_overwhelmed)  # Probability that the server is overwhelmed
```

What is the probability of being overwhelmed at some point during a day? (consider 24hours)

```python
#code here
```

```python
import scipy.stats as stats

# Given values
lambda_visits_per_hour = 500  # Average visits per hour
max_visits_per_hour = 550  # Threshold where server gets overwhelmed
hours_in_day = 24  # Total hours in a day

# Calculate the probability of the server not being overwhelmed in one hour
probability_not_overwhelmed_per_hour = stats.poisson.cdf(max_visits_per_hour, lambda_visits_per_hour)

# Calculate the probability of the server not being overwhelmed in all 24 hours
probability_not_overwhelmed_24_hours = probability_not_overwhelmed_per_hour ** hours_in_day

# Calculate the probability of the server being overwhelmed at least once during the day
probability_overwhelmed_at_least_once = 1 - probability_not_overwhelmed_24_hours

print(probability_overwhelmed_at_least_once)  # Probability that the server is overwhelmed at least once
```

**Challenge 4**

#### Ironhack Helpdesk

Consider a scenario related to the time between arrivals of customers at a service desk.

On average, a customers arrives every 10minutes.

What is the probability that the next customer will arrive within the next 5 minutes?

```python
import math

# Given values
lambda_rate = 1 / 10  # Rate of arrival (1 customer per 10 minutes)
time_period = 5  # Time period we're interested in (5 minutes)

# Calculate the probability using the exponential CDF
probability = 1 - math.exp(-lambda_rate * time_period)

print(probability)  # Probability that the next customer arrives within 5 minutes
```

If there is no customer for 15minutes, employees can that a 5minutes break.

What is the probability an employee taking a break?

```python
import math

# Given values
lambda_rate = 1 / 10  # Rate of arrival (1 customer per 10 minutes)
time_no_customer = 15  # Time period for no customer (15 minutes)

# Calculate the probability using the exponential survival function
probability_no_customer = math.exp(-lambda_rate * time_no_customer)

print(probability_no_customer)  # Probability that no customer arrives in 15 minutes
```

**Challenge 5**

The weights of a certain species of birds follow a normal distribution with a mean weight of 150 grams and a standard deviation of 10 grams. 

- If we randomly select a bird, what is the probability that its weight is between 140 and 160 grams?

```python
#code here
```

```python
import scipy.stats as stats

# Given values
mu = 150  # Mean weight (grams)
sigma = 10  # Standard deviation (grams)

# Z-scores for 140 and 160 grams
z_140 = (140 - mu) / sigma
z_160 = (160 - mu) / sigma

# Calculate the cumulative probabilities
probability = stats.norm.cdf(z_160) - stats.norm.cdf(z_140)

print(probability)  # Probability that the weight is between 140 and 160 grams
```

```python
import math

# Given values
lambda_rate = 1 / 50  # Rate of failure (1 per 50 hours)
time_period = 30  # Time period for failure (30 hours)

# Calculate the probability using the exponential CDF
probability_failure = 1 - math.exp(-lambda_rate * time_period)

print(probability_failure)  # Probability that the component fails within 30 hours
```
