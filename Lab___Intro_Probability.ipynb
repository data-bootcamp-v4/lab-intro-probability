```python
# Challenge 1: Ironhack Airlines
import scipy.stats as st

# Given values
seats = 450
sold_tickets = 460
prob_missing = 0.03
prob_showing_up = 1 - prob_missing

# Define the binomial distribution for the number of passengers showing up
# n = number of trials (sold tickets), p = probability of success (passenger showing up)
# We want the probability that the number of passengers showing up is less than or equal to the number of available seats (450)
n = sold_tickets
p = prob_showing_up

# Calculate the cumulative probability that 450 or fewer passengers show up
prob_all_seats_filled = st.binom.cdf(seats, n, p)

# Print the result
print(f"The probability that there are enough seats for all passengers is {prob_all_seats_filled:.4f}")
```

```python
# Challenge 2: Ironhack Call Center
# Given values
prob_success = 0.3

# Calculate the probability that the representative needs at least 3 attempts
# This is the probability that the first two attempts fail (i.e., no success in the first two trials)
prob_fail_first_two = (1 - prob_success) ** 2

# Print the result
print(f"The probability that the representative needs at least three attempts is {prob_fail_first_two:.4f}")
```

```python
# Challenge 3: Ironhack Website
# Given values
avg_visits_per_hour = 500
server_capacity = 550

# The Poisson distribution is used to model the number of visits per hour
# We want the probability of more than 550 visits (i.e., server being overwhelmed)
prob_overwhelmed = 1 - st.poisson.cdf(server_capacity, avg_visits_per_hour)

# Print the result
print(f"The probability that the website server is overwhelmed is {prob_overwhelmed:.4f}")
```

```python
# Challenge 3: Probability of being overwhelmed during a day (24 hours)
# The probability of the server being overwhelmed in a single hour is calculated above
# The probability that the server is NOT overwhelmed in one hour is the complement
prob_not_overwhelmed_hour = 1 - prob_overwhelmed

# The probability that the server is not overwhelmed in all 24 hours
prob_not_overwhelmed_day = prob_not_overwhelmed_hour ** 24

# The probability that the server is overwhelmed at least once in 24 hours is the complement
prob_overwhelmed_day = 1 - prob_not_overwhelmed_day

# Print the result
print(f"The probability that the server will be overwhelmed at some point during the day is {prob_overwhelmed_day:.4f}")
```

```python
# Challenge 4: Ironhack Helpdesk
# Given values
mean_arrival_time = 10  # Average time between arrivals in minutes
rate = 1 / mean_arrival_time  # Rate parameter (lambda) for the exponential distribution

# 1. Probability that the next customer arrives within the next 5 minutes
prob_next_customer_5_min = 1 - st.expon.cdf(5, scale=mean_arrival_time)

# 2. Probability that no customer arrives for 15 minutes
prob_no_customer_15_min = st.expon.cdf(15, scale=mean_arrival_time)

# Print the results
print(f"The probability that the next customer will arrive within the next 5 minutes is {prob_next_customer_5_min:.4f}")
print(f"The probability that no customer will arrive for 15 minutes (employee can take a break) is {prob_no_customer_15_min:.4f}")
```

```python
# Challenge 5: Bird Weight Distribution
# Given values
mean_weight = 150
std_dev_weight = 10

# Calculate the probability that the bird's weight is between 140 and 160 grams
prob_weight_between_140_160 = st.norm.cdf(160, loc=mean_weight, scale=std_dev_weight) - st.norm.cdf(140, loc=mean_weight, scale=std_dev_weight)

# Print the result
print(f"The probability that the bird's weight is between 140 and 160 grams is {prob_weight_between_140_160:.4f}")
```
